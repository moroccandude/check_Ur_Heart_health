```python
#zenMl
import logging
import os
import pandas as pd
import numpy as np
from sklearn.base import ClassifierMixin
import zenml.integrations.sklearn
from zenml.pipelines import pipeline
from zenml.steps import step
from sklearn.datasets import load_digits
data=load_digits()
data.images[1][1]
```

```python
from sklearn import linear_model
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import r2_score, f1_score, accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
import pandas as pd
import joblib  # Import joblib to save the model
from zenml import step
import numpy as np
from typing_extensions import Annotated

@step
def importer():
    df = pd.read_csv("./source2/heart.csv")
    dependent_var = df[['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal']]
    independent_var = df['target']

    X_train, X_test, Y_train, Y_test = train_test_split(dependent_var, independent_var, test_size=0.4, random_state=42)

    # Return as a tuple
    return X_train, X_test, Y_train, Y_test


@step
def predictor(X_train, X_test, Y_train, Y_test):
    params={}
    # Define the model
    model = DecisionTreeClassifier(params)

    # Train the model
    model.fit(X_train, Y_train)

    # Make predictions
    y_predict = model.predict(X_test)

    # Evaluate model
    f1 = f1_score(Y_test, y_predict)
    print(f"Evaluation: \n F1 score: {f1}")

    accuracy = accuracy_score(Y_test, y_predict)
    print(f"Accuracy: {accuracy:.2f}")

    # Save the trained model to a .pkl file
    # joblib.dump(model, 'decision_tree_model.pkl')
    # print("Model saved to 'decision_tree_model.pkl'")

    return accuracy

@pipeline(enable_cache=False)
def full_pipeline(importer, predictor):
    # Unpack the tuple returned by importer
    X_train, X_test, Y_train, Y_test = importer()
    predictor(X_train, X_test, Y_train, Y_test)

X_train, X_test, Y_train, Y_test = importer()
# Run the pipeline
pipeline = full_pipeline(importer=importer(), predictor=predictor(X_train, X_test, Y_train, Y_test))

```
